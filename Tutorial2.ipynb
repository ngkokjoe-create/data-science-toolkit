3. Create a numpy array of ones and a numpy array of zeroes

```python
one=np.ones((3,3))
zero=np.zeros((3,3))
print(one)
print(zero)
```

4. Create a random numpy array of size 6x3 and transpose it

```python
r_arr=np.random.rand(6,3)
r_arr_trans=r_arr.T
print(r_arr)
print(r_arr_trans)
```

5. Create two random numpy array of size n x p and p x m and then compute the dot product

```python
n=4
p=2
m=3

Arr_1=np.random.rand(n,p)
Arr_2=np.random.rand(p,m)
dot_product=np.dot(Arr_1,Arr_2)

print("Arr_1:")
print(Arr_1)

print("\nArr_2:")
print(Arr_2)

print("\nDot product:")
print(dot_product)
```
